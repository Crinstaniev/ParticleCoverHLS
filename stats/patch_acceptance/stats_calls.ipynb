```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_path = './makepatch_call_counts.csv'
```

```python
df = pd.read_csv(data_path)

df.head()
```

```python
# plot distribution of makepatch_call_count and deletepatch_call_count
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6), dpi=300)
# high resolution plot
sns.set_context("notebook", font_scale=1.5)

sns.violinplot(data=df[['makepatch_call_count', 'deletepatch_call_count']])
sns.despine()

# mark the mean value and max value
plt.axhline(df['makepatch_call_count'].mean(), color='blue', linestyle='dashed', linewidth=2)
plt.axhline(df['deletepatch_call_count'].mean(), color='orange', linestyle='dashed', linewidth=2)

# mark mean value number
plt.text(0, df['makepatch_call_count'].mean(), f"mean: {df['makepatch_call_count'].mean():.2f}", color='blue')
plt.text(1, df['deletepatch_call_count'].mean(), f"mean: {df['deletepatch_call_count'].mean():.2f}", color='orange')

# add space between line and text
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)

# add frame
plt.legend(['makepatch_call_count', 'deletepatch_call_count'])
plt.title('Distribution of makepatch_call_count and deletepatch_call_count')
plt.xlabel('makepatch_call_count and deletepatch_call_count')
plt.ylabel('count')

plt.savefig('makepatch_deletepatch_call_count.png')
plt.show()
```

```python
# calculate ratio of (makepatch_call_count - deletepatch_call_count) / makepatch_call_count with histogram
df['ratio'] = (df['makepatch_call_count'] - df['deletepatch_call_count']) / df['makepatch_call_count']

# plot distribution of ratio
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6), dpi=300)
# high resolution plot
sns.set_context("notebook", font_scale=1.5)

# histogram
sns.histplot(df['ratio'], bins=20, kde=True)
sns.despine()

# mark the mean value and max value
plt.axhline(df['ratio'].mean(), color='blue', linestyle='dashed', linewidth=2)


# add space between line and text
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)

# add frame
plt.legend(['ratio'])
plt.title('Distribution of (makepatch_call_count - deletepatch_call_count) / makepatch_call_count')
plt.xlabel('(makepatch_call_count - deletepatch_call_count) / makepatch_call_count')
plt.ylabel('count')

plt.savefig('makepatch_deletepatch_call_count_ratio.png')
plt.show()
```
